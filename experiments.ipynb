```python
import time
import main

def steps_per_second(step, num_steps):
    saved_pos, saved_vel = main.pos.to_numpy(), main.vel.to_numpy() # both versions are timed from same state
    step() # compile before timing
    main.ti.sync()
    start = time.perf_counter()
    for _ in range(num_steps):
        step()
    main.ti.sync()
    elapsed = time.perf_counter() - start
    main.pos.from_numpy(saved_pos)
    main.vel.from_numpy(saved_vel)
    return num_steps / elapsed

main.init()
for _ in range(round(4.0 / main.config.dt)): # run 4 s of simulation first so particles are mixed, as in a normal run
    main.step_spatially()

naive_sps = steps_per_second(main.step, 200) # GPU acceleration only
print(f"GPU acceleration only: {naive_sps:.2f} steps/s")

optimised_sps = steps_per_second(main.step_spatially, 200) # GPU acceleration + spatial hashing
print(f"GPU acceleration + spatial hashing: {optimised_sps:.2f} steps/s")

time_efficiency_increase = (1 - naive_sps / optimised_sps)
print(f"{(time_efficiency_increase * 100):.2f}% time efficiency increase in optimised solution compared to naive solution")
```
